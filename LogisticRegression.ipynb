```python
import pandas as pd
```

```python
dataset=pd.read_csv("Social_Network_Ads.csv")
```

```python
dataset=pd.get_dummies(dataset,dtype=int,drop_first=True)
dataset
```

```python
dataset=dataset.drop("User ID",axis=1)
```

```python
dataset["Purchased"].value_counts()
```

```python
dataset.columns
```

```python
independent=dataset[['Age', 'EstimatedSalary', 'Gender_Male']]
independent
```

```python
dependent=dataset[['Purchased']]
dependent
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(independent,dependent,test_size=1/3,random_state=0)
```

```python
# Step 1: Logistic Regression calculates z using a linear (straight-line) equation: z = w1*x1 + w2*x2 + b.
# Step 2: This equation combines input features into a single value.
# Step 3: It then applies the sigmoid function: Sigmoid(z) = 1 / (1 + e^(-z)).
# Step 4: Here, e is Euler’s number (approximately 2.718).
# Step 5: The sigmoid function converts z into a probability between 0 and 1.
# Step 6: Finally, this probability is compared with a threshold (usually 0.5) to classify the data into two classes.
from sklearn.linear_model import LogisticRegression
classifier=LogisticRegression(random_state=0)
classifier.fit(x_train,y_train)
```

```python
y_pred=classifier.predict(x_test)
```

```python
from sklearn.metrics import confusion_matrix
confusion_matrix=confusion_matrix(y_test,y_pred)
print(confusion_matrix)
```

```python
pd.DataFrame(confusion_matrix,index=['Predicted Not Purchased(0)','Predicted Purchased(1)'],
             columns=['Actual Not Purchased(0)','Actual Purchased(1)'])
```

```python
from sklearn.metrics import classification_report
classification_report=classification_report(y_test,y_pred)
print(classification_report)
```

```python
result=classifier.predict([[34,10000,1]])
if result[0] == 0:
    print("This person will not purchase")
else:
    print("This person will purchase")

```
